```python
'''# Word Length Dictionary
# Create a dictionary after that convert each sentence into word 
# and modify dictionary with key-value pair for each word 
# and print the dictionary'''

def word_frequency(sentence):
    words = sentence.split()
    d = {}
    for word in words:
        if word not in d:
            d[word] = 1
        else:
            d[word] += 1
    return d
sentence = "python is easy and python is powerful"
result = word_frequency(sentence)
print("Q1 - Result:", result)
```

```python
'''
1. Take a word as input
2. take an empty dict
3. check each character in the word
4. If the character is not in the dict, store it as count 1
5. else increase its count by 1
6. Return and display the dict.'''

# Q2. Find the Frequency of Each Character in a Word
def character_frequency(word):
    frequency = {}

    for char in word:
        if char not in frequency:
            frequency[char] = 1
        else:
            frequency[char] += 1

    return frequency


# Given input
print(character_frequency("banana"))


```

```python
# Q3. Find the First Non-Repeating Character in a Word
def first_count(word):
    list1 = []
    for char in word:
        if word.count(char) == 1:
            list1.append(char)
    if len(list1) == 0:
        return None
    else:
        return list1[0]
print("Q3. Non-Repeating Character:", first_count("swiss"))
```

```python
'''Take two string as input
sort both strings and compare both the string
if they are equal then return True else return False'''

# Q4. Check Whether Two Strings Are Anagrams
def anagrams(s1,s2):
    if sorted(s1) == sorted(s2):
        return True
    else:
        return False
print("Q4. Anagram : ", anagrams("listen","silent"))
```

```python
'''# Q5. Find the First Repeating Character in a String
Take a string as input.create an empty set
Read each character from left to right.
If the character is already in the set
else add the char to the set
If no character repeats, return None '''

def first_repeating_character(string):
    s = set()
    for char in string:
        if char in s:
            return char
        else:
            s.add(char)
    return None
print("Q5. Repeating Character:", first_repeating_character("programming"))

```

```python
'''Move All Zeros
Take empty list
Using for loop in that loop fistly add non zero number using append() method
Then in next loop add the equal to zero waale number then print final list'''


# Q6. Move All Zeros to the End
def move_zeros(number):
    num = []
    for i in number:
        if i != 0:
            num.append(i)
    for i in number:
        if i == 0:
            num.append(i)
    return num
print("Q6. Moved Zeros:", move_zeros([0,1,2,0,4,0,5]))
```

```python
'''# Q7. Find the Second Highest Distinct Number in a List
Take a list of numbers.Remove duplicate numbers
Sort the unique numbers in asce order
If fewer than two unique numbers exist, return None
else return the second-largest number
'''
def second_highest(numbers):
    unique_numbers = sorted(set(numbers))
    if len(unique_numbers) < 2:
        return None
    return unique_numbers[-2]
print(second_highest([10, 20, 40, 30, 40, 20]))

```

```python
'''
Take a list of numbers
Create an empty list
check each number in the original list
If the number is not already in result, add it
Return the result '''


# Q8. Remove Duplicates Without Changing the Original Order
def remove_duplicates(numbers):
    result = []
    for num in numbers:
        if num not in result:
            result.append(num)
    return result

print(remove_duplicates([4, 2, 4, 5, 2, 7, 5]))

```

```python
'''
Take a list as number and the value N.
Calculate the sum of all numbers from 1 to N using N * (N + 1) / 2
Calculate the sum of the numbers present
Subtract the present sum from the total sum'''

# Q9. Find the Missing Number from 1 to N
def find_missing_number(nums, n):
    total = n * (n + 1) // 2
    present_sum = sum(nums)
    return total - present_sum

print(find_missing_number([1, 2, 4, 5, 6], 6))

```

```python
'''
Take a list of numbers and a target value
Select each number from left to right and Check the numbers 
If the sum of a pair equals the target, return that pair
If no pair matches, return an empty list '''


# Q10. Find Two Numbers Whose Sum Equals a Target
def find_pair(nums, target):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return [nums[i], nums[j]]
    return []

print(find_pair([2, 7, 11, 15], 9))

```

```python
'''
Take a list of numbers.
Assume the 1st number is closest to zero and Check each number
if  is smaller update the closest number
If both values are equal choose the +ve number.
Return the closest number '''


# Q11. Find the Number Closest to Zero
def closest_to_zero(nums):
    closest = nums[0]
    for num in nums:
        if abs(num) < abs(closest):
            closest = num
        elif abs(num) == abs(closest) and num > closest:
            closest = num
    return closest

print(closest_to_zero([-8, -3, 2, 5, -2]))
```

```python
'''
take two lists and Create an empty list 
check each element in the first list
Check whether the element exists in the second list and is not already in result.
If both conditions are true, add the element to result.
Return result '''

# Q12. Find Common Elements Between Two Lists
def common_elements(list1, list2):
    result = []
    for num in list1:
        if num in list2 and num not in result:
            result.append(num)

    return result
print("Q12. ", common_elements([1, 2, 3, 4  ], [3, 4, 5, 6]))
```

```python
'''
take a list and the no.of position K
if the list is [], return an [] list
calculate rotations using K the list length
take the last K element and take the remaining elements
Join the last element before the remaining elements
return rotated list '''


# Q13. Rotate a List to the Right by K Positions
def rotate_right(nums, k):
    if len(nums) == 0:
        return []
    k = k % len(nums)
    return nums[-k:] + nums[:-k] if k != 0 else nums.copy()

print(rotate_right([1, 2, 3, 4, 5], 2))
```

```python
'''
take a list having two elements
calculate the sum of the first two elements
check each remaining pair of consecutive element
calculate the sum of the current pair
If the current sum is greater than max sum , update max sum
Return max sum '''


# Q14. Find the Maximum Sum of Two Consecutive Elements
def max_consecutive_sum(nums):
    max_sum = nums[0] + nums[1]

    for i in range(1, len(nums) - 1):
        current_sum = nums[i] + nums[i + 1]

        if current_sum > max_sum:
            max_sum = current_sum

    return max_sum

print(max_consecutive_sum([1,2,3,4,5]))
```

```python
''' Take a list and a target.
create an empty --> pairs
use loop ,if their sum equals  to target, create a pair
sort the pair and add it , if it is not already in pairs ,sort all pairs
print pairs '''


# Q15. Find All Pairs Whose Sum Equals a Target

def find_all_pairs(nums, target):
    pairs = []
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                pair = [nums[i], nums[j]]
                pair.sort()
                if pair not in pairs:
                    pairs.append(pair)
    pairs.sort()
    return pairs

print(find_all_pairs([2, 4, 3, 5, 7, 8, 1], 9))
print(find_all_pairs([1, 2, 3, 4], 5))
```
