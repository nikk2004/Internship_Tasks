```python
# Q1. Find the Shortest Word in a Sentence
# take a sentence as input.
# split the sentence into words.
# find the length of each word.
# select the word with the smallest length.
# and print that word.

sentence = input("Enter a sentence: ")
words = sentence.split()
if words:
    shortest_word = min(words, key=len)
    print("Shortest word:", shortest_word)
else:
    print("The sentence is empty.")
```

```python
# Q2. Find the Largest Number in a List
# take a number as list
# set largest equal to 1st no.
# use a for loop for checking each number.
# if the current number is greater than largest update largest.
# and print result
numbers = [12, 45, 7, 89, 23]
largest = numbers[0]
for num in numbers:
    if num > largest:
        largest = num
print("Largest number:", largest)
```

```python
# Q3. Find the Smallest Number in a List
# create a list of numbers.
# suppose the first number is the smallest
# compare each number with current smallest number.
# if a number is smaller update the smallest number.
# print the result.
numbers = [12, 45, 7, 89, 23]
smallest = numbers[0]
for num in numbers:
    if num < smallest:
        smallest = num
print("Smallest number:", smallest)
```

```python
# # Q4. Find the First Longest Word When Ties Occur
# create a sentence.
# split the sentence into individual words.
# suppose the first word is the longest.
# compare each word with the current longest word.
# if a word is longer, update the longest word.
# if the lengths are equal, keep the first word.
# print the result.

sentence = "Python makes coding useful"
words = sentence.split()
longest = words[0]
for word in words:
    if len(word) > len(longest):
        longest = word
print("First longest word:", longest)
```

```python
# Q5. Find the Word with the Maximum Number of Characters
# create a sentence.
# split the sentence into individual words.
# suppose the first word is the longest.
# compare each word with the current longest word.
# if a word is longer, update the longest word.
# print the result.

sentence = "Data engineering requires problem solving"
words = sentence.split()
longest = words[0]
for word in words:
    if len(word) > len(longest):
        longest = word
print("Word with maximum characters:", longest)
```

```python
# Q6. Count Vowels in a String
# create a string.
# create a variable to store the vowel count.
# check each character in the string.
# if the character is a vowel, increase the count by 1.
# consider both uppercase and lowercase vowels.
# print the total number of vowels.

string = "Education"
count = 0
for char in string:
    if char.lower() in "aeiou":
        count += 1
print("Total no. of vowels:", count)
```

```python
# Q11. Create a Dictionary of Words and Their First Characters
# create a sentence.
# split the sentence into each words.
# create an empty dictionary.
# check each word in the list.
# store each word as a key and its first character as the value.

sentence = "Python Data Engineering"
words = sentence.split()
result = {}
for word in words:
    result[word] = word[0]
print(result)
```

```python
# Q14. Create a Dictionary of Numbers and Their Squares
# create a list of numbers.
# create an empty dictionary.
# check each number in the list.
# store each number as a key and its square as the value.
# print the dictionary.

numbers = [2, 3, 4, 5]
result = {}
for num in numbers:
    result[num] = num ** 2
print(result)
```

```python
# Q15. Create a Dictionary of Numbers and Their Even/Odd Status
# create a list of numbers.
# create an empty dictionary.
# check each number in the list.
# if the number is divisible by 2 store as even value else odd

numbers = [1, 2, 3, 4, 5]
result = {}
for num in numbers:
    if num % 2 == 0:
        result[num] = "Even"
    else:
        result[num] = "Odd"
print(result)
```

```python
# # Q16. Find the Most Frequent Word in a Sentence
# create a sentence.
# split the sentence into individual word
# convert each word to lowercase.
# create an empty dictionary to store word frequencies.
# count word appears how many times
# find the word with the high frequency.
# print the result.

sentence = "Python is easy and Python is powerful"
words = sentence.lower().split()
word_count = {}
for word in words:
    word_count[word] = word_count.get(word, 0) + 1
most_frequent = words[0]
for word in words:
    if word_count[word] > word_count[most_frequent]:
        most_frequent = word
print("Most frequent :", most_frequent)
```

```python
# Q7. Count Consonants in a String
# create a string.
# create a variable to storing consonant word
# check each char in string
# if the char is an alphabet and not a vowel , increase by 1
# ignore spaces, digits, and special char

string = "Python 123!"
count = 0
for char in string:
    if char.isalpha() and char.lower() not in "aeiou":
        count += 1
print("Total number of consonants:", count)
```

```python
# Q8. Find the Most Frequent Character
# create a string.
# remove spaces 
# create an empty dictionary to storing char freq
# count how many times each char appears
# suppose the first char is the most frequent
# compare the frequency of each char with the current most frequent char
# if a char has a high frequency, update the most frequent char
# if frequencies are equal, keep the first char
# print the result.

string = "banana"
string = string.replace(" ", "")
char_count = {}
for char in string:
    char_count[char] = char_count.get(char, 0) + 1
most_frequent = string[0]
for char in string:
    if char_count[char] > char_count[most_frequent]:
        most_frequent = char
print("Most frequent character:", most_frequent)
```

```python
# Q9. Find the Least Frequent Character
# create a string.
# remove spaces from the string.
# create an empty dictionary to store character frequencies.
# count how many times each character appears.
# assume the first character is the least frequent.
# compare the frequency of each character with the current least frequent character.
# if a character has a lower frequency, update the least frequent character.
# if frequencies are equal, keep the first character.
# print the result.

string = "success"
string = string.replace(" ", "")
char_count = {}
for char in string:
    char_count[char] = char_count.get(char, 0) + 1
least_frequent = string[0]
for char in string:
    if char_count[char] < char_count[least_frequent]:
        least_frequent = char
print("Least frequent character:", least_frequent)
```

```python
# Q13. Create a Dictionary of Words and Their Uppercase Versions
# create a sentence
# split the sentence into individual words
# create an empty dictionary
# check each word in the list
# store each word as a key and its uppercase version as the value
# print the dictionary

sentence = "data science python"
words = sentence.split()
result = {}
for word in words:
    result[word] = word.upper()
print(result)
```

```python
# Q12. Create a Dictionary of Words and Their Last Characters
# create a sentence.
# split the sentence into individual word
# create an empty dictionary
# check each word in the list
# store each word as a key and its last character as the value
# print the dictionary

sentence = "Python Data Engineering"
words = sentence.split()
result = {}
for word in words:
    result[word] = word[-1]
print(result)
```

```python
# Q17. Find All Words Appearing More Than Once
# create a sentence.
# convert the sentence to lowercase.
# split the sentence into individual words
# create an empty dictionary to store word frequencies
# count how many times each word appears
# create an empty list to store repeated words
# check each word in the original order.
# print the result

sentence = "data science uses data and science tools"
words = sentence.lower().split()
word_count = {}

for word in words:
    word_count[word] = word_count.get(word, 0) + 1

repeated_words = []

for word in words:
    if word_count[word] > 1 and word not in repeated_words:
        repeated_words.append(word)

print(repeated_words)
```

```python
# Q18. Find the First Word That Appears Only Once
# create a sentence.
# convert the sentence to lowercase
# split the sentence into individual words
# create an empty dictionary to store word frequencies
# count how many times each word appears
# check each word in the original order
# if a word appears only once, select that word
# stop checking after finding the first word that appears only once

sentence = "apple banana apple mango banana orange"
words = sentence.lower().split()
word_count = {}
for word in words:
    word_count[word] = word_count.get(word, 0) + 1
for word in words:
    if word_count[word] == 1:
        print("First non-repeated word:", word)
        break
```

```python
# Q19. Count Words Whose Length Is Greater Than 4
# create a sentence
# split the sentence into individual word
# create a variable to store the count
# check each word in the sentence
# if a word has more than 4 characters, increase the count by 1

sentence = "Python makes problem solving interesting"
words = sentence.split()
count = 0
for word in words:
    if len(word) > 4:
        count += 1
print("Number of words with length greater than 4:", count)
```

```python
# Q20. Find the Longest Word That Appears Only Once
# create a sentence.
# convert each word to lowercase for comparison
# split the sentence into individual words
# create an empty dictionary to store word frequencies
# count how many times each word appears
# assume the longest qualifying word is empty
# check each word in the original order
# if a word appears only once and is longer, update the longest word
# if lengths are equal, keep the first word

sentence = "Python makes coding useful Python"
words = sentence.split()
word_count = {}
for word in words:
    word_lower = word.lower()
    word_count[word_lower] = word_count.get(word_lower, 0) + 1
longest = ""
for word in words:
    if word_count[word.lower()] == 1:
        if len(word) > len(longest):
            longest = word
print("Longest word that appears only once:", longest)
```
